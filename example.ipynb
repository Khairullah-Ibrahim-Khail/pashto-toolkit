# Pashto Toolkit

Fake Afghan data in Pashto script (`pa_AF`) and Latin transliteration (`en_AF`).

**Kernel:** pick `Python (Pashto Toolkit)` or `.venv/bin/python` from the kernel selector, top right.


```python
from pashto_toolkit import PashtoFaker

pa = PashtoFaker("pa_AF")   # Pashto script
en = PashtoFaker("en_AF")   # Latin transliteration

print(len(pa.formatters()), "formatters available")
```

## One person

```python
print("name     :", pa.name())
print("job      :", pa.job())
print("province :", pa.province())
print("phone    :", pa.phone_number())
print("tazkira  :", pa.afghan_id())
print("bank     :", pa.bank_name())
print("price    :", pa.pricetag())
```

## A table of people

Run the cell again for fresh data.

```python
for _ in range(10):
    print(f"{pa.name():<30} {pa.province():<14} {pa.phone_number():<18} {pa.afghan_id()}")
```

## Both locales side by side

```python
for field in ("name", "province", "city", "bank_name", "month_name", "day_of_week", "currency_name", "word"):
    print(f"{field:<14} {pa.format(field):<28} {en.format(field)}")
```

## Reproducible data

Pass a `seed` and every run gives the same values. That is what makes it usable in tests.

```python
a = PashtoFaker("pa_AF", seed=42)
b = PashtoFaker("pa_AF", seed=42)

print([a.name() for _ in range(3)])
print([b.name() for _ in range(3)])
print("identical:", [PashtoFaker('pa_AF', seed=1).name()] == [PashtoFaker('pa_AF', seed=1).name()])
```

## A whole record

```python
import json

print(json.dumps(pa.profile(), indent=2, ensure_ascii=False, default=str))
```

## Straight into pandas

`json()` emits rows ready to load. Install pandas first if you want this cell:
`.venv/bin/pip install pandas`

```python
try:
    import pandas as pd

    rows = json.loads(pa.json(
        data_columns=[("name", "name"), ("province", "province"),
                      ("phone", "phone_number"), ("tazkira", "afghan_id")],
        num_rows=12,
    ))
    display(pd.DataFrame(rows))
except ImportError:
    print("pandas is not installed; run:  .venv/bin/pip install pandas")
```
